```python
#Funcion para llamar a la API
def obtener_info_artista(nombre):
    url = "http://ws.audioscrobbler.com/2.0/"
    
    params = {
        "method": "artist.getinfo",
        "artist": nombre,
        "api_key": API_KEY,
        "format": "json"
    }
    
    response = requests.get(url, params=params)
    return response.json()
```

```python
#funcion para limpiar datos
def procesar_artista(nombre):
    data = obtener_info_artista(nombre)
    
    try:
        artista = data["artist"]
        
        bio = artista["bio"]["summary"]
        listeners = artista["stats"]["listeners"]
        playcount = artista["stats"]["playcount"]
        
        similares = [a["name"] for a in artista["similar"]["artist"]]
        
        return {
            "artista": nombre,
            "biografia": bio,
            "listeners": listeners,
            "playcount": playcount,
            "similares": ", ".join(similares)
        }
    
    except:
        return None
```

```python
#Bucle para todos los artistas
resultados = []

for artista in artistas:
    info = procesar_artista(artista)
    if info:
        resultados.append(info)
```
